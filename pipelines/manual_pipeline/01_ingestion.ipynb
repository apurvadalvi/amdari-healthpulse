Python script to ingest data in Snowflake

```python
import pandas as pd
import json
from snowflake.connector.pandas_tools import write_pandas
from snowflake.connector import connect
```

```python
df = pd.read_parquet('data/healthpulsedataset.parquet')
print(df.head)
```

```python
print(df.dtypes)
```

```python
# Load credentials from config.json
with open("config.json", "r") as f:
    config = json.load(f)
```

```python
# Connect to Snowflake
conn = connect(
    user= config['user'],
    password = config['password'],
    account = config['account'],
    warehouse = config['warehouse'],
    database = config['database'],
    schema = config['schema'],
    role = config['role']
)
try:
    # Push DataFrame to Snowflake (creates or replaces table)
    success, nchunks, nrows, _ = write_pandas(conn, df, 'raw_data', auto_create_table = True)
    print(f"Success: {success}, Rows Loaded: {nrows}")
    print('done')
except Exception as e:
    print(e)
    print("Couldnt load file")
```
