```python
# import libraries
import pandas as pd
```

```python
# read the input file
df = pd.read_parquet('../data/healthpulsedataset.parquet')
```

```python
# initialise batch size
batch_size = 2000
```

```python
# split input file into batches and save in parquet format
for i in range(0, len(df), batch_size):
    batch_df = df.iloc[i:i+batch_size]
    batch_df.to_parquet(f'../data/batched_data/healthpulse_batch_{i//batch_size}.parquet')
```
